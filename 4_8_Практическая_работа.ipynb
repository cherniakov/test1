### Код

```python
import pandas as pd
from datetime import datetime

def gorokh_prepare(path):
  df = pd.read_csv(path)
  df = df.rename(columns={
  'Месяц': 'month',
  'Квартал': 'quarter',
  'Год': 'year',
  'РФ, средняя цена': 'price'}
  )
  df['price_date'] = df[['year', 'month']].apply(lambda row: datetime(*row, 1), axis=1)
  return df[['month', 'quarter', 'year', 'price_date','price']].to_dict(orient='records')

gorokh_path = 'горох.csv'
data = gorokh_prepare(gorokh_path)
```

```python
# df = pd.read_csv(gorokh_path, index_col=0)
# df.head()
```

```python
data = pd.DataFrame(data)
data.head()
```

```python
plt.subplots(figsize=(15, 8))
plt.plot(data['price_date'], data['price'])

plt.title('Цены на горох')
plt.ylabel('Цена')
plt.xlabel('Дата')
plt.xticks(data['price_date'][::3], rotation=45)
plt.show()
```
